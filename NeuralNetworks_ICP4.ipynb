```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam

# Define path to CSV file
csv_file_path = 'diabetes.csv'

# Load dataset
diabetes_data = pd.read_csv(csv_file_path, header=None).values

# Split dataset into training and test sets
X_train, X_test, y_train, y_test = train_test_split(
    diabetes_data[:, 0:8], diabetes_data[:, 8],
    test_size=0.25, random_state=87
)

# Set random seed for reproducibility
np.random.seed(155)

# Create a simple neural network model
simple_nn_model = Sequential()
simple_nn_model.add(Dense(20, input_dim=8, activation='relu'))
simple_nn_model.add(Dense(1, activation='sigmoid'))

# Compile the model
simple_nn_model.compile(loss='binary_crossentropy', optimizer=Adam(), metrics=['accuracy'])

# Train the model
simple_nn_model.fit(
    X_train, y_train, epochs=100, initial_epoch=0
)

# Print model summary
print(simple_nn_model.summary())

# Evaluate the model
test_loss, test_accuracy = simple_nn_model.evaluate(X_test, y_test)
print(f"Test Loss: {test_loss:.4f}")
print(f"Test Accuracy: {test_accuracy:.4f}")

# Create a deeper neural network model
deep_nn_model = Sequential()
deep_nn_model.add(Dense(64, input_dim=8, activation='relu'))  # First hidden layer
deep_nn_model.add(Dense(32, activation='relu'))  # Second hidden layer
deep_nn_model.add(Dense(16, activation='relu'))  # Third hidden layer
deep_nn_model.add(Dense(8, activation='relu'))  # Fourth hidden layer
deep_nn_model.add(Dense(1, activation='sigmoid'))  # Output layer

# Compile the deeper model
deep_nn_model.compile(loss='binary_crossentropy', optimizer=Adam(), metrics=['accuracy'])

# Train the deeper model
deep_nn_model.fit(
    X_train, y_train, epochs=100, initial_epoch=0
)

# Print model summary
print(deep_nn_model.summary())

# Evaluate the deeper model
deep_test_loss, deep_test_accuracy = deep_nn_model.evaluate(X_test, y_test)
print(f"Test Loss (Deeper Model): {deep_test_loss:.4f}")
print(f"Test Accuracy (Deeper Model): {deep_test_accuracy:.4f}")
```

1. Plot the loss and accuracy for both training data and validation data using the history object in the source code.

```python
from keras import Sequential
from keras.datasets import mnist
import numpy as np
from keras.layers import Dense
from keras.utils import to_categorical
import matplotlib.pyplot as plt

# Load and process the MNIST dataset
(train_imgs, train_lbls), (test_imgs, test_lbls) = mnist.load_data()

# Reshape and normalize the dataset
num_pixels = np.prod(train_imgs.shape[1:])
train_imgs_reshaped = train_imgs.reshape(train_imgs.shape[0], num_pixels).astype('float32') / 255.0
test_imgs_reshaped = test_imgs.reshape(test_imgs.shape[0], num_pixels).astype('float32') / 255.0

# Convert labels to one-hot encoding
train_lbls_one_hot = to_categorical(train_lbls)
test_lbls_one_hot = to_categorical(test_lbls)

# Create the neural network model
mnist_model = Sequential([
    Dense(512, activation='relu', input_shape=(num_pixels,)),
    Dense(512, activation='relu'),
    Dense(10, activation='softmax')
])

# Compile the model
mnist_model.compile(optimizer='rmsprop', loss='categorical_crossentropy', metrics=['accuracy'])

# Train the model
training_history = mnist_model.fit(
    train_imgs_reshaped, train_lbls_one_hot,
    batch_size=256, epochs=10, verbose=1,
    validation_data=(test_imgs_reshaped, test_lbls_one_hot)
)

# Extract training history data
history_data = training_history.history
epochs_range = range(1, len(history_data['accuracy']) + 1)

# Plot training & validation accuracy
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.plot(epochs_range, history_data['accuracy'], 'bo', label='Training Accuracy')
plt.plot(epochs_range, history_data['val_accuracy'], 'b', label='Validation Accuracy')
plt.title('Training and Validation Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()

# Plot training & validation loss
plt.subplot(1, 2, 2)
plt.plot(epochs_range, history_data['loss'], 'bo', label='Training Loss')
plt.plot(epochs_range, history_data['val_loss'], 'b', label='Validation Loss')
plt.title('Training and Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()

plt.tight_layout()
plt.show()
```

The program imports keras, numpy, matplotlib.pyplot, and mnist dataset for building and training a neural network.

Reads the MNIST dataset, which contains handwritten digit images (28x28 pixels) and their corresponding labels (0-9).

Converts 28x28 images into a 1D array (784 pixels) and scales pixel values to a range of 0 to 1 for better training performance.

Converts the digit labels (0-9) into categorical format, making it easier for the neural network to learn.

Defines a 3-layer deep learning model with:

Two hidden layers (each with 512 neurons, ReLU activation).
One output layer (10 neurons, Softmax activation) to classify digits from 0 to 9.
Uses categorical_crossentropy as the loss function and rmsprop as the optimizer to adjust weights for better accuracy.

The model is trained for 10 epochs using a batch size of 256, learning from training images and validating against test images.
Stores the model’s accuracy and loss values for both training and validation phases across epochs.


A graph is plotted showing Training vs. Validation Accuracy over epochs.
Another graph shows Training vs. Validation Loss, helping to analyze performance trends.
The plots help visualize how well the model is learning and whether it’s overfitting or underfitting.
